```python
tokens = word_tokenize(document.lower().replace(',', '').replace('·', ''))
```

```python
#vocab
vocab={'<unk>':0}
for token in Counter(tokens).keys():
    if token not in vocab:
        vocab[token]=len(vocab)
```

```python
def text_to_indices(sentences,vocab):
    numerical_sentences=[]
    
    for token in sentences:
        if token in vocab:
            numerical_sentences.append(vocab[token])
        else:
            numerical_sentences.append(vocab['<unk>'])
            
    return numerical_sentences
```

```python
input_numerical_sentences=[]
for sentences in input_sentences:
    
    input_numerical_sentences.append(text_to_indices(word_tokenize(sentences.lower()),vocab))
```

```python
from typing import Any

from torch.utils.data import DataLoader,Dataset

class CustomDataset(Dataset):
    def __init__(self,X,y):
        self.X=X
        self.y=y
    
    def __len__(self):
        return self.X.shape[0]
    
    def __getitem__(self, index):
        
        return self.X[index],self.y[index]
```

```python
dataset=CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataLoader=DataLoader(dataset,batch_size=32,shuffle=True)
```

```python
for input,output in dataLoader:
    print(input,output)
```

```python
class LSTM(nn.Module):
    def __init__(self,vocab_size):
        super().__init__()
        self.embedding=nn.Embedding(vocab_size,100)
        self.lstm=nn.LSTM(100,150,batch_first=True)
        self.fc=nn.Linear(150,vocab_size)
        
    def forward(self,X):
       emdedded= self.embedding(X)
       intermediate_hidden_state,(final_hidden_state,final_cell_state)=self.lstm(emdedded)
       output=self.fc(final_hidden_state.squeeze(0))
       return output
       
```

```python
model=LSTM(len(vocab))
```

```python
device=torch.device('cuda'if torch.cuda.is_available() else 'cpu')
```

```python
model.to(device)
```

```python
epochs=50
lr=0.001
criterion=nn.CrossEntropyLoss()
optimizer=torch.optim.Adam(model.parameters(),lr=lr)
```

```python
for epoch in range(epochs):
    total_loss=0
    
    for batch_x,batch_y in dataLoader:
        batch_x,batch_y=batch_x.to(device),batch_y.to(device)
        optimizer.zero_grad()
        output=model(batch_x)
        loss=criterion(output,batch_y)
        
        loss.backward()
        
        optimizer.step()
        
        total_loss=total_loss + loss.item()
        
    print(f'Epochs: {epoch+1},Loss:{total_loss:.4f}')
        
        
```

```python
#prediction 

def prediction(model,vocab,text):
    #tokenzie
    tokenized_text=word_tokenize(text.lower())
    
    #text into numerical indices
    numerical_text=text_to_indices(tokenized_text,vocab)
    
    #padding
    padded_text=torch.tensor([0]*(90 - len(numerical_text)) + numerical_text,dtype=torch.long).unsqueeze(0)
    
    #send to model 
    output=model(padded_text)
    
    #predicted  word index
    value,index=torch.max(output,dim=1)
    
    #merge with text 
    return text + ' '  + list(vocab.keys())[index]
        
    
    
```
